# Python

Python è oggi uno dei linguaggi di programmazione più diffusi e viene utilizzato in numerosi ambiti, dallo sviluppo software alla Data Science, dal Machine Learning al calcolo scientifico. La sua popolarità è dovuta soprattutto alla sintassi chiara, alla grande disponibilità di librerie specializzate e a una comunità di sviluppatori molto ampia.

In questo corso daremo per acquisita una conoscenza di base del funzionamento e della sintassi di Python. Tuttavia, prima di affrontare le librerie e gli strumenti specifici per il calcolo numerico, in questo notebook richiameremo brevemente alcuni elementi fondamentali del linguaggio.

In particolare, rivedremo la definizione e l’uso delle variabili, i principali tipi di dato, gli operatori, le strutture condizionali e iterative, le principali strutture dati e la definizione di funzioni.

Questa introduzione non vuole essere un corso completo di programmazione in Python, ma un ripasso delle conoscenze necessarie per comprendere e svolgere le successive attività di laboratorio. Chi non ha sufficiente familiarità con la sintassi di base è invitato a integrare questo materiale con un tutorial completo.


## Premessa

In Python l’*indentazione* non ha soltanto una funzione grafica, ma fa parte della sintassi del linguaggio: viene utilizzata per identificare le istruzioni appartenenti allo stesso blocco di codice. Convenzionalmente, ogni livello di indentazione corrisponde a quattro spazi.

Un’indentazione mancante o incoerente produce un errore oppure modifica il significato del programma.
